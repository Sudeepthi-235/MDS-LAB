# NumPy Array Indexing (First Element as 101)

```python
import numpy as np
```

## Create a Sample NumPy Array

```python
# Create a sample NumPy array with the first element as 101
arr = np.array([[101, 2, 3], [4, 5, 6], [7, 8, 9]])
print("Original Array:\n", arr)
```

## Indexing Examples

```python
# Accessing a single element (first element)
print("Element at (0, 0):", arr[0, 0])
```

```python
# Accessing another single element
print("Element at (1, 2):", arr[1, 2])
```

```python
# Accessing the first row
print("First row:", arr[0, :])
```

```python
# Accessing the last column
print("Last column:\n", arr[:, -1])
```

## Boolean Indexing

```python
# Accessing elements using boolean indexing
bool_arr = arr > 5
print("Boolean array (arr > 5):\n", bool_arr)
print("Elements greater than 5:\n", arr[bool_arr])
```

## Slicing NumPy Arrays

```python
# Slicing: First two rows, first two columns
print("First two rows, first two columns:\n", arr[:2, :2])
```

```python
# Slicing: All rows, middle column
print("All rows, middle column:\n", arr[:, 1:2])
```

```python
# Slicing: Last row, last two columns
print("Last row, last two columns:\n", arr[-1, -2:])
```

## Joining NumPy Arrays

```python
# Create two arrays for joining, setting the first element of arr1 to 101
arr1_join = np.array([[101, 2], [3, 4]])
arr2_join = np.array([[5, 6], [7, 8]])

print("Array 1 for joining:\n", arr1_join)
print("Array 2 for joining:\n", arr2_join)
```

```python
# Concatenate along rows (axis=0)
arr_joined_row = np.concatenate((arr1_join, arr2_join), axis=0)
print("Joined along rows (axis=0):\n", arr_joined_row)
```

```python
# Concatenate along columns (axis=1)
arr_joined_col = np.concatenate((arr1_join, arr2_join), axis=1)
print("Joined along columns (axis=1):\n", arr_joined_col)
```

```python
# Stacking vertically (vstack)
arr_vstack = np.vstack((arr1_join, arr2_join))
print("Stacked vertically (vstack):\n", arr_vstack)
```

```python
# Stacking horizontally (hstack)
arr_hstack = np.hstack((arr1_join, arr2_join))
print("Stacked horizontally (hstack):\n", arr_hstack)
```

## Computation on NumPy Arrays using Universal Functions (UFuncs) and Mathematical Methods

```python
# Create an array for computations, ensuring one element is 101
comp_arr = np.array([-2, -1, 0, 101, 2, 3, 4, 5, 6])
print("Original Array for computations:", comp_arr)
```

### Universal Functions (UFuncs)

```python
# Universal Function: Absolute values (np.abs)
print("Absolute values (np.abs):", np.abs(comp_arr))
```

```python
# Universal Function: Square root (np.sqrt) - using an array with non-negative values for illustration
# We ensure one element is 101 for the "first element as 101" requirement
sqrt_arr = np.array([101.0, 4.0, 9.0, 16.0])
print("Square root (np.sqrt):", np.sqrt(sqrt_arr))
```

```python
# Universal Function: Exponential (np.exp)
print("Exponential (np.exp):", np.exp(np.array([0, 1, 2])))
```

```python
# Universal Function: Sine (np.sin)
print("Sine (np.sin):", np.sin(comp_arr))
```

### Mathematical Methods

```python
# Mathematical Method: Sum of all elements (np.sum)
print("Sum of all elements (np.sum):", np.sum(comp_arr))
```

```python
# Mathematical Method: Mean of elements (np.mean)
print("Mean of elements (np.mean):", np.mean(comp_arr))
```

```python
# Mathematical Method: Standard deviation (np.std)
print("Standard deviation (np.std):", np.std(comp_arr))
```

```python
# Mathematical Method: Maximum element (np.max)
print("Maximum element (np.max):", np.max(comp_arr))
```

```python
# Mathematical Method: Minimum element (np.min)
print("Minimum element (np.min):", np.min(comp_arr))
```

```python
# Matrix Multiplication (np.dot)
matrix_mult_arr1 = np.array([[101, 2], [3, 4]]) # First element set to 101
matrix_mult_arr2 = np.array([[5, 6], [7, 8]])
print("Matrix 1:\n", matrix_mult_arr1)
print("Matrix 2:\n", matrix_mult_arr2)
print("Matrix Multiplication (np.dot):\n", np.dot(matrix_mult_arr1, matrix_mult_arr2))
```

## Operations using Random Functions

```python
# Generate a 3x3 array of random floats between 0 and 1
random_float_array = np.random.rand(3, 3)
print("Random 3x3 float array:\n", random_float_array)
```

```python
# Generate a 2x4 array of random integers between 1 and 100
random_int_array = np.random.randint(1, 101, size=(2, 4)) # Example with 101 as upper bound (exclusive)
print("Random 2x4 integer array (1-100):\n", random_int_array)
```

```python
# Generate a 2x2 array from a normal distribution with mean 0 and standard deviation 1
random_normal_array = np.random.normal(loc=0.0, scale=1.0, size=(2, 2))
print("Random 2x2 array from normal distribution:\n", random_normal_array)
```
