```python
import pandas as pd
import numpy as np

#Load the datatset from the CSV file
df=pd.read_csv('/content/hospital_patients_real_world.csv')
# Display the first few rows of the DataFrame
display(df.head())
# Print the shape (number of rows and columns) of the DataFrame
print(df.shape)
```

```python
print(df.isnull().sum()) # Calculate and display the count of missing values for each column in the DataFrame.
```

```python
# Fill missing 'Age' values with the median age
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing 'Gender' values with the mode (most frequent) gender
df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

# Fill missing 'Diagnosis' values with the mode (most frequent) diagnosis
df['Diagnosis'] = df['Diagnosis'].fillna(df['Diagnosis'].mode()[0])

# Print a message indicating that missing values have been filled
print("Missing values after filling:")
# Display the count of missing values for each column after filling
print(df.isnull().sum())
```

```python
df_drop = pd.read_csv("hospital_patients_real_world.csv")

print("Before dropping:", df_drop.shape)

df_drop = df_drop.dropna()

print("After dropping:", df_drop.shape)
```

```python
Q1 = df['Age'].quantile(0.25) # Calculate the first quartile (25th percentile) of 'Age'
Q3 = df['Age'].quantile(0.75) # Calculate the third quartile (75th percentile) of 'Age'

IQR = Q3 - Q1 # Calculate the Interquartile Range (IQR)

lower_bound = Q1 - 1.5 * IQR # Calculate the lower bound for outlier detection
upper_bound = Q3 + 1.5 * IQR # Calculate the upper bound for outlier detection

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
outliers = df[
    (df['Age'] < lower_bound) | # Select rows where 'Age' is below the lower bound
    (df['Age'] > upper_bound)  # Select rows where 'Age' is above the upper bound
]

print("Number of outliers:", len(outliers)) # Print the total count of identified outliers
print(outliers) # Display the DataFrame containing the outliers
```

```python
df_clean = df[
    (df['Age'] >= lower_bound) & # Filter rows where 'Age' is greater than or equal to the lower bound
    (df['Age'] <= upper_bound)  # Filter rows where 'Age' is less than or equal to the upper bound
]

print("Original rows:", len(df)) # Print the number of rows in the original DataFrame
print("Rows after outlier filtering:", len(df_clean)) # Print the number of rows after removing outliers
```
