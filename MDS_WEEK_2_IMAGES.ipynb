```python
from PIL import Image
import numpy as np

# Load the image
img_path = '/content/DEER.jpg'
img = Image.open(img_path)

# Convert the image to a NumPy array
img_array = np.array(img)

print(f"Image shape: {img_array.shape}")
print(f"Image data type: {img_array.dtype}")
print("First 5x5 pixels (RGB values):\n", img_array[:5, :5, :])
```

```python

import cv2
import matplotlib.pyplot as plt

# Ensure img_path is available
img_path = '/content/DEER.jpg'

# Load the image using OpenCV (BGR format by default)
img_cv2_original_bgr = cv2.imread(img_path)

# Convert original BGR image to RGB for display with matplotlib
img_cv2_original_rgb = cv2.cvtColor(img_cv2_original_bgr, cv2.COLOR_BGR2RGB)

# 1. Horizontal Flip using OpenCV
img_cv2_flipped_horizontal = cv2.flip(img_cv2_original_bgr, 1) # 1 for horizontal flip
img_cv2_flipped_horizontal_rgb = cv2.cvtColor(img_cv2_flipped_horizontal, cv2.COLOR_BGR2RGB)

# 2. Vertical Flip using OpenCV
img_cv2_flipped_vertical = cv2.flip(img_cv2_original_bgr, 0) # 0 for vertical flip
img_cv2_flipped_vertical_rgb = cv2.cvtColor(img_cv2_flipped_vertical, cv2.COLOR_BGR2RGB)

# 3. Rotate 90 degrees clockwise using OpenCV
img_cv2_rotated_90 = cv2.rotate(img_cv2_original_bgr, cv2.ROTATE_90_CLOCKWISE)
img_cv2_rotated_90_rgb = cv2.cvtColor(img_cv2_rotated_90, cv2.COLOR_BGR2RGB)

# 4. Rotate 270 degrees clockwise using OpenCV
img_cv2_rotated_270 = cv2.rotate(img_cv2_original_bgr, cv2.ROTATE_90_COUNTERCLOCKWISE) # This is equivalent to 270 clockwise
img_cv2_rotated_270_rgb = cv2.cvtColor(img_cv2_rotated_270, cv2.COLOR_BGR2RGB)

# Create a figure and a set of subplots
fig, axes = plt.subplots(2, 3, figsize=(18, 12))

# Display Original Image
axes[0, 0].imshow(img_cv2_original_rgb)
axes[0, 0].set_title('Original Image (CV2)')
axes[0, 0].axis('off')

# Display Horizontally Flipped Image
axes[0, 1].imshow(img_cv2_flipped_horizontal_rgb)
axes[0, 1].set_title('Horizontally Flipped (CV2)')
axes[0, 1].axis('off')

# Display Vertically Flipped Image
axes[0, 2].imshow(img_cv2_flipped_vertical_rgb)
axes[0, 2].set_title('Vertically Flipped (CV2)')
axes[0, 2].axis('off')

# Display 90-degree Rotated Image
axes[1, 0].imshow(img_cv2_rotated_90_rgb)
axes[1, 0].set_title('Rotated 90 Degrees (CV2)')
axes[1, 0].axis('off')

# Display 270-degree Rotated Image
axes[1, 1].imshow(img_cv2_rotated_270_rgb)
axes[1, 1].set_title('Rotated 270 Degrees (CV2)')
axes[1, 1].axis('off')

# Hide the empty subplot
fig.delaxes(axes[1, 2])

plt.tight_layout()
plt.show()
```

## Student Marks Data Analysis

Let's perform various statistical and comparison operations on the student marks data from the `Simple 100 Student Marks.csv` file.

```python
import pandas as pd

# Load the CSV file
csv_path = '/content/Simple 100 Student Marks.csv'
df_marks = pd.read_csv(csv_path)

print("First 5 rows of the dataset:")
display(df_marks.head())
print("\n")

print("Columns in the dataset:", df_marks.columns.tolist())

# Define the marks columns for iteration
marks_columns = ['maths_marks', 'science_marks', 'english_marks', 'social_studies_marks', 'language_marks']
```

### Descriptive Statistics for Each Subject

```python
for col in marks_columns:
    if col in df_marks.columns:
        print(f"Descriptive statistics for {col.replace('_', ' ').title()}:")
        display(df_marks[col].describe())
        print("\n" + "-" * 50 + "\n") # Separator for better readability
    else:
        print(f"Column '{col}' not found in the dataset.")
```

### Students with Marks Above 80 in Each Subject

```python
for col in marks_columns:
    if col in df_marks.columns:
        high_achievers = df_marks[df_marks[col] > 80]
        print(f"Students with {col.replace('_', ' ').title()} > 80:")
        if not high_achievers.empty:
            display(high_achievers[['student_name', col]])
        else:
            print(f"No students scored above 80 in {col.replace('_', ' ').title()}.")
        print("\n" + "-" * 50 + "\n") # Separator for better readability
```

### Student(s) with the Maximum Mark in Each Subject

```python
for col in marks_columns:
    if col in df_marks.columns:
        max_mark = df_marks[col].max()
        top_students = df_marks[df_marks[col] == max_mark]
        print(f"Student(s) with the highest {col.replace('_', ' ').title()} ({max_mark}):")
        display(top_students[['student_name', col]])
        print("\n" + "-" * 50 + "\n") # Separator for better readability
```

### Student(s) with the Minimum Mark in Each Subject

```python
for col in marks_columns:
    if col in df_marks.columns:
        min_mark = df_marks[col].min()
        lowest_students = df_marks[df_marks[col] == min_mark]
        print(f"Student(s) with the lowest {col.replace('_', ' ').title()} ({min_mark}):")
        display(lowest_students[['student_name', col]])
        print("\n" + "-" * 50 + "\n") # Separator for better readability
```
