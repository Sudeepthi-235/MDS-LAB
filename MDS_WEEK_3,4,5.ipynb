```python
import pandas as pd
```

```python
df = pd.read_csv('/content/Most Streamed Artists on Spotify (17_07_2026) V1.1.csv')
```

```python
df.head(10)
```

```python
df.tail(10)
```

```python
print(f"Shape of the DataFrame: {df.shape}")
print(f"Index of the DataFrame: {df.index}")
print(f"Columns of the DataFrame: {df.columns}")
```

```python
# Select rows where 'Country of Origin' is 'United States'
print("\n--- Artists from United States ---\n")
df_us_artists = df[df['Country of Origin'] == 'United States']
print(df_us_artists.head())

# Delete rows where 'Total Streams (in millions)' is less than 10,000
print("\n--- DataFrame after deleting artists with low streams ---\n")
df_high_streams = df[df['Total Streams (in millions)'] >= 10000]
print(df_high_streams.head())
print(f"Original DataFrame shape: {df.shape}, New DataFrame shape after deleting rows: {df_high_streams.shape}")

# Select specific columns
print("\n--- Selecting 'Artist Name' and 'Primary Genre' columns ---\n")
df_subset_columns = df[['Artist Name', 'Primary Genre']]
print(df_subset_columns.head())

# Delete a column (e.g., 'Primary Language')
print("\n--- DataFrame after deleting 'Primary Language' column ---\n")
df_dropped_column = df.drop(columns=['Primary Language'])
print(df_dropped_column.head())
print(f"Original DataFrame columns: {df.columns.tolist()}\nNew DataFrame columns after dropping: {df_dropped_column.columns.tolist()}")
```

```python
# Sort the DataFrame by 'Total Streams (in millions)' in descending order
print("\n--- DataFrame sorted by Total Streams (Descending) ---\n")
df_sorted_desc = df.sort_values(by='Total Streams (in millions)', ascending=False)
print(df_sorted_desc[['Artist Name', 'Total Streams (in millions)']].head())

# Sort the DataFrame by 'Total Streams (in millions)' in ascending order
print("\n--- DataFrame sorted by Total Streams (Ascending) ---\n")
df_sorted_asc = df.sort_values(by='Total Streams (in millions)', ascending=True)
print(df_sorted_asc[['Artist Name', 'Total Streams (in millions)']].head())

# Add a rank column based on 'Total Streams (in millions)' (highest streams get rank 1)
print("\n--- DataFrame with 'Total Streams' Rank ---\n")
df['Total Streams Rank'] = df['Total Streams (in millions)'].rank(ascending=False, method='min')
print(df[['Artist Name', 'Total Streams (in millions)', 'Total Streams Rank']].sort_values(by='Total Streams Rank').head())
```

```python
# Statistical operations on 'Total Streams (in millions)'
print("\n--- Statistics for 'Total Streams (in millions)' ---\n")
print(f"Mean: {df['Total Streams (in millions)'].mean():,.2f}")
print(f"Median: {df['Total Streams (in millions)'].median():,.2f}")
print(f"Minimum: {df['Total Streams (in millions)'].min():,.2f}")
print(f"Maximum: {df['Total Streams (in millions)'].max():,.2f}")
print(f"Standard Deviation: {df['Total Streams (in millions)'].std():,.2f}")

# Statistical operations on 'Debut Year'
print("\n--- Statistics for 'Debut Year' ---\n")
print(f"Mean: {df['Debut Year'].mean():.2f}")
print(f"Median: {df['Debut Year'].median():.2f}")
print(f"Minimum: {df['Debut Year'].min():.0f}")
print(f"Maximum: {df['Debut Year'].max():.0f}")
print(f"Standard Deviation: {df['Debut Year'].std():.2f}")

# Display full descriptive statistics for selected numerical columns
print("\n--- Descriptive Statistics for key numerical columns ---\n")
print(df[['Total Streams (in millions)', 'Debut Year', '% of Solo Streams']].describe())
```

```python
categorical_columns = ['Sex', 'Country of Origin', 'Primary Language', 'Primary Genre', ' Artist Type']

for col in categorical_columns:
    print(f"\n--- Analysis for '{col.strip()}' ---\n")
    # Number of unique values
    unique_count = df[col].nunique()
    print(f"Number of unique values: {unique_count}")

    # Count of each unique value
    value_counts = df[col].value_counts()
    print("Value counts:\n", value_counts)
```

```python
print("Original columns:", df.columns.tolist())

# Rename a single column (e.g., ' Artist Type' to 'Artist Type Clean')
df_renamed_single = df.rename(columns={' Artist Type': 'Artist Type Clean'})
print("\n--- After renaming ' Artist Type' to 'Artist Type Clean' ---")
print(df_renamed_single.columns.tolist())

# Rename multiple columns (e.g., 'Sex' to 'Gender' and 'Debut Year' to 'First Album Year')
df_renamed_multiple = df.rename(columns={'Sex': 'Gender', 'Debut Year': 'First Album Year'})
print("\n--- After renaming 'Sex' and 'Debut Year' ---")
print(df_renamed_multiple.columns.tolist())

# To apply changes directly to the original DataFrame, you can use inplace=True or reassign:
# df.rename(columns={'OldName': 'NewName'}, inplace=True)
# or
# df = df.rename(columns={'OldName': 'NewName'})
```

```python
import pandas as pd

# 1. Create a Pandas Series from a Python List
data_list = [10, 20, 30, 40, 50]
series_from_list = pd.Series(data_list)
print("--- Pandas Series from a List ---")
print(series_from_list)
```

```python
# 2. Create a Pandas Series from a Python Dictionary (keys become index)
data_dict_series = {'a': 100, 'b': 200, 'c': 300}
series_from_dict = pd.Series(data_dict_series)
print("\n--- Pandas Series from a Dictionary ---")
print(series_from_dict)
```

```python
# 3. Create a Pandas DataFrame from a Dictionary of Lists (column-wise)
data_dict_df = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston']
}
df_from_dict = pd.DataFrame(data_dict_df)
print("\n--- Pandas DataFrame from a Dictionary of Lists (Column-wise) ---")
print(df_from_dict)
```

```python
# 4. Create a Pandas DataFrame from a List of Dictionaries (row-wise)
data_list_of_dicts = [
    {'Name': 'Eve', 'Age': 22, 'City': 'Miami'},
    {'Name': 'Frank', 'Age': 28, 'City': 'Seattle'},
    {'Name': 'Grace', 'Age': 33, 'City': 'Boston'}
]
df_from_list_of_dicts = pd.DataFrame(data_list_of_dicts)
print("\n--- Pandas DataFrame from a List of Dictionaries (Row-wise) ---")
print(df_from_list_of_dicts)
```
